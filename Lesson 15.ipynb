# HANDS-ON EXERCISE 1
## Business Analytics — Sales Performance Calculator

```python
import numpy as np
```

```python
# Task 1: Create the Dataset
daily_sales = np.array([125000, 150000, 175000, 140000, 190000, 210000, 160000])
print("Daily Sales:", daily_sales)
print("Type:", type(daily_sales))
print("Shape:", daily_sales.shape)

print()

# Task 2: Calculate Total Sales
total_sales = np.sum(daily_sales)
print("Total Sales for the Week: ₦", total_sales)

print()

# Task 3: Calculate Average Daily Sales
average_sales = np.mean(daily_sales)
print("Average Daily Sales: ₦", average_sales)
print("Average Daily Sales: ₦", round(average_sales, 2))

print()

# Task 4: Apply a 10% Sales Adjustment
# adjusted_sales = daily_sales * (1 + 10/100)
adjusted_sales = daily_sales * 1.10
print("Adjusted Sales:", adjusted_sales)

print()

# Task 5: Compare Original and Adjusted Sales
additional_revenue = adjusted_sales - daily_sales
print("Additional Revenue per Day:", additional_revenue)

print()

# Task 6: Use Numpy's Mathematical Functions(Compare Average and Total Sales)
print("Average Daily Sales: ₦", round(average_sales, 2))
print("Total Weekly Sales: ₦", total_sales)
```

## Average Daily Sales vs Total Sales 
The total sales figure shows the overall revenue generated by the business during the week, while the average daily sales shows the typical amount of revenue generated per day. The average helps management understand daily sales performance and provides a useful benchmark for comparing individual days.

# HANDS-ON EXERCISE 2
## Education Analytics — Student Performance Analysis


```python
# Task 1: Create and Inspect the Data
scores = np.array([62, 75, 81, 69, 88, 94, 73, 85, 77, 91])
print("Scores:", scores)
print("Type:", type(scores))
print("Shape:", scores.shape)

print()


# Task 2: Calculate the Mean Score
mean_score = np.mean(scores)
print("Mean Score:", mean_score)

print()


# Task 3: Calculate Each Student's Deviation
deviations = scores - mean_score
print("Deviations:", deviations)

print()


# Task 4: Identify Positive and Negative Deviations (Understanding Deviations)
# A positive deviation means that a student's score is above the group average.
# A negative deviation means that a student's score is below the group average.
# A deviation close to zero means that the student's score is close to the group average.

print()


# Task 5: Square the Deviations
squared_deviations = deviations ** 2
print("Squared Deviations:", squared_deviations)


print()


# Task 6: Calculate the Sum of Squared Deviations
sum_squared_deviations = np.sum(squared_deviations)
print("Sum of Squared Deviations:", sum_squared_deviations)
```

## Why Do We Square the Deviations?
Squaring the deviations changes all negative values into positive values because multiplying a negative number by itself produces a positive number. This prevents positive and negative deviations from cancelling each other out when we calculate the overall variation in the scores.

## Standard Deviation Process
The standard deviation calculation can be understood through these steps:

1. Calculate the mean.
2. Calculate the deviation of each score from the mean.
3. Square the deviations.
4. Sum the squared deviations.
5. Divide by the appropriate number of observations.
6. Take the square root.

Standard deviation is useful because it shows how spread out the students' scores are around the average. While the mean tells us the typical score, standard deviation gives us additional information about whether the scores are closely grouped around the mean or widely spread out.

## Industry Thinking Challenge
The average score gives an overall picture of student performance, but it does not show how individual scores are distributed around that average. Looking at deviations and standard deviation helps the coordinator identify whether students are performing at similar levels or whether there are large differences in performance that may require further attention.

## Engineering & Scientific Computing — Trigonometric Series and Convergence

```python
# Task 1: Define the angle
theta = 30
theta_radians = np.radians(theta)
print("Theta in degrees:", theta)
print("Theta in radians:", theta_radians)

print()


# Task 2: Create the sequence
k = np.arange(1, 100, 1)
print("First value:", k[0])
print("Last value:", k[-1])
print("Number of values:", len(k))

print()


# Task 3: Calculate the trigonometric component
sin_theta = np.sin(theta_radians)
print("sin(theta):", sin_theta)

print()


# Task 4: Build the series terms
series_terms = sin_theta / k
print("Series Terms:", series_terms)

print()


# Task 5: Calculate the sum
series_sum = np.sum(series_terms)
print("Series Sum:", series_sum)

print()


# Task 6: Increase the number of terms
k_99 = np.arange(1, 100, 1)
series_99 = np.sum(sin_theta / k_99)

k_999 = np.arange(1, 1000, 1)
series_999 = np.sum(sin_theta / k_999)

k_9999 = np.arange(1, 10000, 1)
series_9999 = np.sum(sin_theta / k_9999)

print("99 terms:", series_99)
print("999 terms:", series_999)
print("9,999 terms:", series_9999) 
```

# Task 7: Investigate Convergence
| Number of Terms | Series Result |
|---:|---:|
| 99 | 2.58869 |
| 999 | 3.74224 |
| 9,999 | 4.89375 |
The result increases as more terms are added

# Task 8: Understanding Convergence
In this experiment, increasing the number of terms causes the series result to increase. With 99 terms, the result is approximately 2.58869, while 999 terms gives approximately 3.74224 and 9,999 terms gives approximately 4.89375.

For a convergent series, increasing the number of terms would cause the result to approach a fixed limiting value. However, the series used in this exercise is proportional to the harmonic series, which does not converge. Therefore, the result continues to increase as more terms are added rather than approaching a fixed limit.

# BONUS CHALLENGE- Food Production Performance Analysis

### Food Production Performance Analysis Using NumPy
This mini-project uses NumPy to analyze daily food production output over one week.

### Scenario
A food processing company records the number of food packages produced each day.
The production manager wants to understand the company's weekly production,
average daily output, and how much production would increase if output improved
by 10%.

### Objective

The objective of this experiment is to use NumPy to calculate the total and
average production, apply a 10% production increase, and determine the
additional number of packages that could be produced.

### Dataset
The following dataset represents the number of food packages produced by the
company over seven days. The data is simulated for this numerical experiment.

```python
# Dataset
daily_production = np.array([1200, 1350, 1180, 1450, 1500, 1320, 1250])

print()

# Inspect the array
print("Daily Production:", daily_production)
print("Type:", type(daily_production))
print("Shape:", daily_production.shape)

print()

# Calculate total production
total_production = np.sum(daily_production)
print("Total Weekly Production:", total_production)

print()

# Calculate average production
average_production = np.mean(daily_production)
print("Average Daily Production:", round(average_production, 2))

print()

# Apply 10% increase
adjusted_production = daily_production * (1 + 10/100)
print("Adjusted Production:", adjusted_production)

print()

# Calculate additional production
additional_production = adjusted_production - daily_production
print("Additional Production:", additional_production)

print()

# Calculate total adjusted production
total_adjusted_production = np.sum(adjusted_production)
print("Original Weekly Production:", total_production)
print("Adjusted Weekly Production:", total_adjusted_production)
```

### Results

| Measurement | Result |
|---|---:|
| Total weekly production | 9,250 packages |
| Average daily production | 1,321.43 packages |
| Production after 10% increase | 10,175 packages |
| Additional weekly production | 925 packages |

### Interpretation

The company produced a total of 9,250 packages during the week, with an average
daily production of approximately 1,321 packages. A hypothetical 10% increase
in production would raise weekly output to 10,175 packages, giving the company
925 additional packages for the week.

The average daily production provides a useful benchmark for understanding the
company's normal daily output, while the total production shows the overall
weekly output. The adjusted figures also help management estimate the potential
effect of improving production efficiency.

### Conclusion

NumPy can be used to quickly analyze production data and perform calculations
on an entire array at once. This experiment demonstrated how numerical tools
can help a food production company understand its output and estimate the
potential impact of increased production.
